```python
import numpy as np
import cv2
import matplotlib.pyplot as plt
import matplotlib.cm as cm
from PIL import Image

print("--------------------------------------")

# Task 1: Loading with OpenCV
img = cv2.imread("images/cameraman.tif")
plt.imshow(img)
plt.show()

print("--------------------------------------")

# Task 2: Loading with PIL
img2 = Image.open("images/cameraman.tif")
# Manual requires Greys_r for this step
plt.imshow(img2, cmap=cm.Greys_r) 
plt.show()

print("--------------------------------------")

# Task 3: Saving Images
cv2.imwrite("new_image.jpg", img)
img2.save("new_image2.jpg")

# Task 4: Array Shapes and Values
print(img.shape)
print("--------------------------------------")
print(img)
print("--------------------------------------")

img_array = np.array(img2)
print(img_array.shape)
print("--------------------------------------")
print(img_array)
print("--------------------------------------")
```
