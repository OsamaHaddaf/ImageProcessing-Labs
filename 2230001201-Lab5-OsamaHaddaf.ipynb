```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Read the image and convert to float32 for processing
img = cv2.imread('images/cameraman.tif', 0)
img_float = img.astype(np.float32) / 255.0

# --- Assessment Task #1: 7x7 Box Filter ---
# Convolve image with a box filter for smoothing
box_blur = cv2.blur(img, (7, 7))

plt.figure(figsize=(10, 5))
plt.subplot(1, 2, 1); plt.imshow(img, cmap='gray'); plt.title('Original')
plt.subplot(1, 2, 2); plt.imshow(box_blur, cmap='gray'); plt.title('7x7 Box Filter')
plt.show()

# --- Assessment Task #2: Gaussian Filters (5x5 and 21x21) ---
# Apply Gaussian filters with different kernel sizes to show varying blur
gauss_5 = cv2.GaussianBlur(img, (5, 5), 0)
gauss_21 = cv2.GaussianBlur(img, (21, 21), 0)

plt.figure(figsize=(15, 5))
plt.subplot(1, 3, 1); plt.imshow(img, cmap='gray'); plt.title('Original')
plt.subplot(1, 3, 2); plt.imshow(gauss_5, cmap='gray'); plt.title('5x5 Gaussian')
plt.subplot(1, 3, 3); plt.imshow(gauss_21, cmap='gray'); plt.title('21x21 Gaussian')
plt.show()

# --- Assessment Task #3: Laplacian Sharpening ---
# 1. Use the cv2.Laplacian() function
laplacian = cv2.Laplacian(img_float, cv2.CV_32F)

# 2. Sharpen the image using the formula from manual
sharpened = np.clip(img_float - laplacian, 0, 1)

# 3. Plot original, filter result, and sharpened image
plt.figure(figsize=(15, 5))
plt.subplot(1, 3, 1); plt.imshow(img_float, cmap='gray'); plt.title('Original')
plt.subplot(1, 3, 2); plt.imshow(laplacian, cmap='gray'); plt.title('Laplacian Filter')
plt.subplot(1, 3, 3); plt.imshow(sharpened, cmap='gray'); plt.title('Sharpened Image')
plt.show()

# Save a result for the lab folder
cv2.imwrite('images/lab5_sharpened.jpg', (sharpened * 255).astype(np.uint8))
```
