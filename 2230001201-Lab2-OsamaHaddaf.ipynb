```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

print("--------------------------------------")

# Functions for sampling and quantization
def sample_image(image, factor):
    h, w = image.shape[:2]
    # Resize down then up to simulate sampling
    small = cv2.resize(image, (w // factor, h // factor), interpolation=cv2.INTER_NEAREST)
    return cv2.resize(small, (w, h), interpolation=cv2.INTER_NEAREST)

def quantize_image(image, levels):
    step = 256 // levels
    return (np.floor(image / step) * step).astype(np.uint8)

print("--------------------------------------")

# Task 1: Sampling & Quantization
# Loading image
img = cv2.imread('images/cameraman.tif', 0) 

# Applying sampling and quantization
sampled = sample_image(img, 10)
quantized = quantize_image(img, 4)

# Displaying Task 1 results
plt.figure(figsize=(10, 4))
plt.subplot(131); plt.imshow(img, cmap='gray'); plt.title('Original')
plt.subplot(132); plt.imshow(sampled, cmap='gray'); plt.title('Sampled')
plt.subplot(133); plt.imshow(quantized, cmap='gray'); plt.title('Quantized')
plt.show()

print("--------------------------------------")

# Task 2: Arithmetic & Logical Operations
# Loading images (Updated to use 'moon.tif')
img1 = cv2.imread('images/cameraman.tif', 0)
img2 = cv2.imread('images/moon.tif', 0)  # <--- Changed this line

# Resizing to ensure they match (Required for math operations)
img1 = cv2.resize(img1, (256, 256))
img2 = cv2.resize(img2, (256, 256))

# Arithmetic Operations
# Subtraction
sub_img = cv2.subtract(img1, img2)

# Add constant 175
add_const = cv2.add(img1, 175)

# Set Operations
# Set difference (img1 - img2)
set_diff = cv2.bitwise_and(img1, cv2.bitwise_not(img2))

# Symmetric difference (XOR)
sym_diff = cv2.bitwise_xor(img1, img2)

# Intersection (AND)
intersection = cv2.bitwise_and(img1, img2)

# Displaying Task 2 results
plt.figure(figsize=(12, 6))
plt.subplot(231); plt.imshow(sub_img, cmap='gray'); plt.title('Subtraction')
plt.subplot(232); plt.imshow(add_const, cmap='gray'); plt.title('Add 175')
plt.subplot(233); plt.imshow(set_diff, cmap='gray'); plt.title('Set Difference')
plt.subplot(234); plt.imshow(sym_diff, cmap='gray'); plt.title('Symmetric Diff')
plt.subplot(235); plt.imshow(intersection, cmap='gray'); plt.title('Intersection')
plt.tight_layout()
plt.show()

print("--------------------------------------")
```
